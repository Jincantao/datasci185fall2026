**Question (A) - (material from Lecture 9)** Counting coffee orders with a dictionary

The set-up cell creates a random list `orders`, where each element is the drink one customer ordered.

1. Using a `for` loop, build a dictionary `drink_counts` whose keys are the drinks that appear in `orders` and whose values are how many times each one was ordered. Start from an **empty** dictionary. Drinks that nobody ordered should **not** appear as keys.
2. Loop over `drink_counts` and print one line per drink in the form `latte: 3 orders`
3. Using a **running maximum** (like Lecture 9, Exercise 2), find the most popular drink and print `Most popular: latte (3 orders)`. If there is a tie, any of the tied drinks is fine.

You may **not** use the list method `.count()` or any library that counts for you.

````python
###### Set-up code (do not change)
import random

menu = ["latte", "americano", "cold brew", "matcha", "mocha", "chai"]
orders = random.choices(menu, k = random.randint(8, 15))
print(orders)
###### (end of set-up code)
````

**Question (E) - (material from Lecture 12)** Cleaning typed survey responses

Students answered the survey question "How many hours did you sleep last night?" by typing into a web form, so every response is a **string** - except that skipped questions were stored as `None`.

Write a function `clean_sleep(responses)` that takes a list like `responses` and **returns two values**: a list of the valid answers converted to **integers**, and the number of responses that were skipped. A response is valid only if it is a **whole number** between 0 and 24 (both inclusive). For each response that is skipped, print a message:

- If the response is `None`: print `Missing response`
- If the response is not a whole number (e.g. `"six"`, `"7.5"`, `""`): print `Sorry, six isn't a whole number` (with the response in place of `six`)
- If it is a whole number outside 0 to 24: print `Sorry, 30 is out of range`

You must detect the first two problems with `try` / `except` (catch the **specific** error types - no bare `except:`), and the third with an `if` statement.

**Example:**
```python
valid, skipped = clean_sleep(["7", None, "six", "30", "8 ", "7.5"])
```
should print `Missing response`, `Sorry, six isn't a whole number`, `Sorry, 30 is out of range`, `Sorry, 7.5 isn't a whole number`, and give `valid == [7, 8]` and `skipped == 4`.

Then run your function on the random `responses` from the set-up cell and print `Average sleep: 7.25 hours (4 skipped)` (average rounded to 2 decimal places).

````python
###### Set-up code (do not change)
import random

possible = ["7", "8", "6", "9", "5", "10", "six", "7.5", "", None, "30", "-1", " 8", "eight", "24", "0"]
responses = random.sample(possible, k = random.randint(6, 10))
responses.append("7")     # guarantees at least one valid answer
print(responses)
###### (end of set-up code)
````

````python
# # your code here
import numpy as np


def clean_sleep(response):
    valid = []
    skipped = 0
    for answer in response:

        if answer == None:
            print(f"Missing Response")
            skipped += 1
            continue



        try:
            int(answer)
            if 0<= int(answer) <= 24:
                valid.append(int(answer))
                continue

            else:
                print(f"sorry {answer} is out of range")
                skipped += 1
                continue


        except ValueError:
            print(f"sorry {answer} is not a whole number")
            skipped += 1

            


    return valid, skipped

    
sleep,skip = clean_sleep(responses)
            
print(f" Average sleep: {round ( np.mean(sleep) ,2)} ({skip} skipped)")
````
