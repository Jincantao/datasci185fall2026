**Question (A) - (pandas: random splits and `.query`)** Splitting app users into three test groups

- Define an integer variable `n`, the number of rows of `df_users`, using `len` or `.shape`.

- Use `.query` to split the users into **three** DataFrames:
    - `pilot`: users with `random_var` less than 0.25
    - `beta`: users with `random_var` at least 0.25 but less than 0.6
    - `control`: users with `random_var` at least 0.6
- Use `assert` to check that the three groups together contain exactly `n` users.
- Print one line per group showing the **percentage** of users in that group, rounded to 1 decimal place, in exactly this form:

```text
pilot: 24.7% of 231 users

control: 40.3% of 231 users
```

(Your numbers will be different, but they should be close to 25%, 35%, and 40%.)

````python
###### Set-up code (do not change)
import numpy as np
import pandas as pd
import random

num_users = random.randint(150, 300)
df_users = pd.DataFrame({"user_id": range(1, num_users + 1),
                         "minutes_per_day": np.random.randint(5, 120, num_users)})
df_users.head()
###### (end of set-up code)
````

**Question (D) - (OOP: a method that changes an object)** Phone battery

The cell below defines a `PhoneBattery` class. Each battery has a `name`, a `charge` between 0 and 100 (percent), and a count `num_sessions` of how many times an app was used. The class already has getters and a setter `set_charge` that uses `assert` to keep the charge between 0 and 100.

**Your task:** add a method `use_app(self, minutes)` that

- uses `assert` to make sure `minutes` is **greater than 0**, with the message `"minutes must be positive"`,
- adds 1 to `num_sessions`, and
- lowers the charge by `0.4` per minute, but **never below 0**. Use the setter to change the charge, rather than changing the attribute directly.

**For example:**

```python
phone = PhoneBattery("Pixel", 80)
phone.use_app(30)
print(phone)

old = PhoneBattery("Old", 10)
old.use_app(60)
print(old)
```

should print `Pixel: 68.0% (1 sessions)` and then `Old: 0% (1 sessions)` (`0.0%` is also fine), not an `AssertionError`.

After the test cell, run the cell below it. It should fail with your `assert` message.

````python
# Add the method use_app to the PhoneBattery class below

class PhoneBattery:
    """A phone battery that drains when apps are used."""

    def __init__(self, name, charge):
        self.name = name
        self.charge = charge
        self.num_sessions = 0

    def __str__(self):
        return f"{self.name}: {round(self.charge, 1)}% ({self.num_sessions} sessions)"

    # ---------- getters ----------
    def get_name(self):
        return self.name

    def get_charge(self):
        return self.charge

    def get_num_sessions(self):
        return self.num_sessions

    # ---------- setter ----------
    def set_charge(self, value):
        assert 0 <= value <= 100, "charge must be between 0 and 100"
        self.charge = value

    # ***************************************************
    # ---------- your code here: write use_app ----------
    # ***************************************************

    def use_app(self,minutes):
        assert minutes > 0, "minutes must be positive"

        self.num_sessions += 1
        if self.charge - (0.4 * minutes) > 0:
            self.set_charge(self.charge - (0.4*minutes))
        elif self.charge - (0.4 * minutes) <= 0:
            self.set_charge(0)
````

*This question builds on your `PhoneBattery` class from (D). Make sure the (D) test cell works first.*

A "pro" battery only drains a fraction of what a normal battery would. Write a class `ProBattery` that inherits from `PhoneBattery` and:

1. has a **class variable** `efficiency` equal to `0.75`,
2. overrides `use_app(self, minutes)` so that it calls the **parent's** `use_app` with `minutes * efficiency` (use `super()`; don't copy the parent's code),
3. overrides `__str__` so that it returns `"Pro "` followed by the parent's string (again use `super()`), e.g. `Pro Galaxy: 71.0% (1 sessions)`,
4. has a **class method** `set_efficiency(cls, value)` that uses `assert` to make sure `value` is between 0.5 and 1 (0.5 and 1 are allowed), then changes `efficiency` for **every** pro battery.

After the test cell, run the two cells below it. Each should fail with your `assert` message.

**Part 2.** Answer in the markdown cell at the end: *without running it*, what does the last cell of this question print, and why are the three numbers not all the same?

````python
# Define the ProBattery class here

class ProBattery (PhoneBattery):
    efficiency = 0.75
    def use_app(self,minutes):
        act_min = minutes * self.efficiency
        super().use_app(act_min)

    def __str__(self):
        return f"Pro {super().__str__()}"

    @classmethod
    def set_efficiency(cls,value):
        assert 0.5 <= value <= 1
        cls.efficiency = value
        
````

**Question (F) - (OOP: objects working together, a loop inside a loop)** The tutoring center

The set-up cell defines a class `Tutor` (do not change it). Each tutor has a `name`, a `subject`, and a list `students` (starting empty), with getters and a method `add_student(student_name)`. It also creates a random list `requests` of `(student, subject)` tuples.

**Your task:** write a class `TutorCenter` with

- a **class variable** `max_students` equal to `2`: a tutor can't take more students than this.
- a constructor `__init__(self, tutor_dict)`, where `tutor_dict` maps each tutor's name to their subject (like `{"Maya": "calculus", "Leo": "python"}`). Loop over the dictionary and store a `Tutor` object for each tutor in a list attribute called `tutors`.
- a method `assign(self, requests)` that goes through `requests` **in order**. For each `(student, subject)`, find the **first** tutor in `tutors` who teaches that subject **and** has fewer than `max_students` students, and use `add_student` to give them that student. If no tutor fits, put the student's name in a list called `waitlist`. The method should **return** `waitlist`.

*Hint 1:* you'll need a `for` loop inside a `for` loop. *Hint 2:* to stop at the **first** tutor who fits, you can use `break`, or a variable like `found = False`.

Then, in the last code cell, set `TutorCenter.max_students = 3`, build a **new** center with the same test tutors, and run `assign` on the same test requests. Your code should give the second expected output **without changing anything inside your class**.

````python
###### Set-up code (do not change)
import random

class Tutor:
    """A peer tutor who can teach one subject."""

    def __init__(self, name, subject):
        self.name = name
        self.subject = subject
        self.students = []

    def __str__(self):
        return f"{self.name} ({self.subject}): {self.students}"

    def get_name(self):
        return self.name

    def get_subject(self):
        return self.subject

    def get_students(self):
        return self.students

    def add_student(self, student_name):
        self.students.append(student_name)

tutor_dict = {"Maya": "calculus", "Leo": "python", "Zoe": "calculus", "Sam": "chemistry"}
names = ["Ana", "Ben", "Cy", "Dee", "Eli", "Fay", "Gus", "Hal", "Ivy", "Jo"]
requests = [(name, random.choice(["calculus", "python", "chemistry", "biology"]))
            for name in random.sample(names, k = random.randint(5, 10))]
print("requests:", requests)
###### (end of set-up code)

type(requests[0])
requests[0][0]


for i in tutor_dict:
    print(i)
````

````python
# Define the TutorCenter class here

class TutorCenter():
    max_students = 2
    def __init__(self,tutor_dict):
        self.tutors = []
        for tutor in tutor_dict:
            self.tutors.append(Tutor(tutor,tutor_dict[tutor]))

    def assign(self, requests):
        waitlist = []
        for i in range(len(requests)):
            sub = requests[i][1]
            name = requests[i][0]
            for tutor in self.tutors:
                a = tutor.get_subject()
                if a == sub and len(tutor.get_students()) < self.max_students:
                    tutor.add_student(name)

                    break
            else:                     
                waitlist.append(name)
        return waitlist


            
````
